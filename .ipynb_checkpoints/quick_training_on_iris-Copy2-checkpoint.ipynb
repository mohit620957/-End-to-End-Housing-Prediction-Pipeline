```python
import pandas as pd
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
data = pd.read_csv("iris-train.csv")
```

```python
data.head()
```

```python
x = data.iloc[:, :-1]
```

```python
x
```

```python
y = data.iloc[:,-1]
```

```python
y
```

```python
model = RandomForestClassifier()
```

```python
x.values
```

```python
model.fit(x.values,y.values) # model got trained finally
```

```python
predictions = model.predict([[5.1, 3.5, 1.4, 0.2]])
```

```python
predictions
```

```python
datatest = pd.read_csv("iris-test.csv")
```

```python
x_test= datatest.iloc[:,:-1]
```

```python
predictions = model.predict(x_test.values)
```

```python
predictions
```
